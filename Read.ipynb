```python
import pandas as pd
df=pd.read_csv(r"C:\Users\adity\OneDrive\Documents\New folder\project\diabetes.csv")
```

```python
#replacing value of insulin
p = int(df['Insulin'].mean()//1)
df['Insulin'] = df['Insulin'].replace(0, p)
```

```python
#replacing value of glucose
p = int(df['Glucose'].mean()//1)
df['Glucose'] = df['Glucose'].replace(0, p)
```

```python
#replacing value of SkinThickness
p = int(df['SkinThickness'].mean()//1)
df['SkinThickness'] = df['SkinThickness'].replace(0, p)
```

```python
p=int(df['BloodPressure'].mean()//1)
df['BloodPressure'] = df['BloodPressure'].replace(0, p)
```

```python
p=int(df["BMI"].mean()//1)
df["BMI"]=df["BMI"].replace(0, p)
```

```python
df.to_csv(r"C:\Users\adity\OneDrive\Documents\New folder\project\diabetes_modified.csv", index=False)
```
