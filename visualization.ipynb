```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("diabetes_modified.csv")

sns.countplot(x='Outcome', data=df)
plt.title("Diabetes Outcome Distribution")
plt.show()
"""this shows the outcomes of the diabetes dataset, where 0 represents non-diabetic and 1 represents diabetic.
500 people in the dataset are non-diabetic, while 268 people are diabetic.
which makes the dataset imbalanced, as there are more non-diabetic cases than diabetic cases."""
```

```python
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
"""this shows the distribution of each feature in the diabetes dataset."""

```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)
plt.title("Boxplot for Outlier Detection")
plt.show()
```

```python
sns.boxplot(x='Outcome', y='Glucose', data=df)
plt.title("Glucose Levels vs Diabetes Outcome")
plt.show()
```

```python
sns.boxplot(x='Outcome', y='BMI', data=df)
plt.title("BMI vs Diabetes Outcome")
plt.show()
```

```python
sns.histplot(data=df, x="Age", hue="Outcome", kde=True)
plt.title("Age Distribution by Outcome")
plt.show()
```

```python
sns.pairplot(df, hue='Outcome')
plt.show()
```

```python
sns.scatterplot(x='Glucose', y='BMI', hue='Outcome', data=df)
plt.title("Glucose vs BMI")
plt.show()
```
